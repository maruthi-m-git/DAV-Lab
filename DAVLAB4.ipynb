```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("students.csv")
```

```python
print("First 5 Records:")
print(df.head())
```

```python
plt.figure(figsize=(8, 5))

sns.histplot(
    df["Marks"],
    bins=10
)
plt.title("Distribution of student Marks")
plt.xlabel("Marks")
plt.ylabel("Number of students")

plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(
    x=df["Marks"]
)
plt.title("Box Plot of student Marks")
plt.xlabel("Marks")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(
    x=df["Branch"]
)
plt.title("Box Plot of student Branch")
plt.xlabel("Branch")
plt.show()
```

```python
plt.figure(figsize=(8,4))
sns.boxplot(
    x=df["StudyHours"]
)
plt.title("Box Plot of student StudyHours")
plt.xlabel("StudyHours")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

sns.scatterplot(
    x="StudyHours",
    y="Marks",
    data=df,
    hue="Result",
    palette={
        "Pass": "green",
        "Fail": "red"
    },
    s=70
)

sns.regplot(
    x="StudyHours",
    y="Marks",
    data=df,
    scatter=False
)
```

```python
sns.pairplot(df[["StudyHours","Attendance","PrevMarks","Marks"]])
plt.show()
```

```python
correlation=df[["StudyHours","Attendance","PrevMarks","Marks"]].corr()
print("\nCorrelation Matrix:\n")
print(correlation)
```

```python
plt.figure(figsize=(8,6))
sns.heatmap(correlation,annot=True)
plt.title("Correlation Heatmap")
plt.show()
```
